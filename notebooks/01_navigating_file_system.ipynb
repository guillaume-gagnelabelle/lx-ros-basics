# Navigation dans le système de fichiers

Ce tutoriel présente les concepts du système de fichiers de ROS et explique comment utiliser les commandes `roscd`, `rosls` et `rospack`.

## Aperçu des concepts liés aux systèmes de fichiers

- **Packages:** Les packages sont l'unité d'organisation logicielle du code ROS. Chaque packages peut contenir des libraries, des exécutables, des scripts ou d'autres éléments.
- **Manifests (fichiers package.xml):** Un manifest est une description d'un paquet logiciel. Il définit les dépendances entre les packages et contient des méta-informations sur le packages lui-même, telles que la version, le responsable, la licence, etc.

## Outils du système de fichiers

Le code source est généralement réparti sur de nombreux packages ROS, même pour les petits projets. Naviguer avec des outils de ligne de commande traditionnels tels que `ls` et `cd` peut s'avérer très fastidieux. C'est pourquoi ROS propose des outils pour faciliter la navigation dans le système de fichiers.

### `rospack`

La commande `rospack` permet d'obtenir des informations sur les packages. Dans ce tutoriel, nous n'aborderons que l'option `find`, qui renvoie le chemin d'accès au paquet souhaité.

#### Usage:

```{bash}
    rospack find [package_name]
```

Par exemple, en exécutant la commande,

```{bash}
    rospack find roscpp
```

reviendrait:

    /opt/ros/noetic/share/roscpp


### `roscd`

La commande `roscd` permet de changer de répertoire (de manière similaire à la commande `cd` de Linux) pour accéder directement à un package.

#### Usage:

```{bash}
    roscd [package_name]
```

Pour vérifier que nous avons bien basculé vers le répertoire du package `roscpp`, exécutez cet exemple :

```{bash}
    roscd roscpp
```

Affichons maintenant le répertoire de travail à l'aide de la commande Linux `pwd` :

```{bash}
    pwd
```

Vous devriez voir :

    /opt/ros/noetic/share/roscpp

Vous pouvez constater que `/opt/ros/noetic/share/roscpp` est le même chemin que celui indiqué par la commande `rospack find` dans l'exemple précédent.


Notez que `roscd`, comme les autres outils ROS, ne trouvera que les paquets ROS se trouvant dans les répertoires listés dans votre variable d'environnement `ROS_PACKAGE_PATH`. Pour afficher le contenu de votre variable `ROS_PACKAGE_PATH`, exécutez la commande suivante :

```{bash}
    echo ${ROS_PACKAGE_PATH}
```

Votre variable d'environnement `ROS_PACKAGE_PATH` doit contenir une liste de répertoires où se trouvent vos packages ROS, séparés par des deux-points `:`.

De manière similaire à d'autres variables d'environnement, vous pouvez ajouter des répertoires supplémentaires à votre variable `ROS_PACKAGE_PATH`, chaque chemin étant séparé par un deux-points `:`.



#### Sous-répertoires

La commande `roscd` permet également de se déplacer vers un sous-répertoire d'un package.

Essayez :

```{bash}
    roscd roscpp/cmake
    pwd
```

Vous devriez voir :

    /opt/ros/noetic/share/roscpp/cmake

## `rosls`

La commande `rosls` vous permet d'exécuter la commande Linux `ls` directement dans un package en le désignant par son nom plutôt que par son chemin absolu.

#### Usage:

```{bash}
    rosls [package_name]
```

Par exemple,

```{bash}
    rosls roscpp_tutorials
```

reviendrait,

    cmake launch package.xml  srv

## Sommaire

Vous avez peut-être remarqué une certaine logique dans la manière dont ces commandes ROS sont nommées :

- `rospack` = **ros** + `pack`(age)
- `roscd` = **ros** + `cd`
- `rosls` = **ros** + `ls`

Ce schéma de dénomination est valable pour de nombreux outils ROS.


Passez au [notebook suivant consacré aux Catkin workspaces](../notebooks/02_catkin_workspaces.ipynb).
