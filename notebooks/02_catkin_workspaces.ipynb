# Catkin Workspaces

Un Catkin workspace est un dossier dans lequel vous modifiez, compilez et installez des packages ROS (catkin).

Un Catkin workspace typique contient au moins quatre répertoires, : `src/`, `build/`, `logs/`, `devel/`, chacun jouant un rôle différent dans le processus de développement logiciel.

### `src`

Le répertoire `src/` contient le code source organisé en Catkin packages. C'est là que vous ajoutez le code source des packages que vous souhaitez compiler. Chaque dossier de cet workspace contient un ou plusieurs  Catkin packages.

### `build/`

Le répertoire `build/` est l'endroit où Catkin stocke ses informations de cache et autres fichiers intermédiaires résultant du processus build (ou de compilation) d'un espace de travail Catkin.

### `logs/`

Le répertoire `logs/` est l'endroit où Catkin stocke les journaux du processus de compilation. Ces journaux contiennent tous les outputs `stdout` et `stderr` générés pendant la compilation. Ces données sont utiles pour le débogage des erreurs de compilation.

### `devel/`

Le répertoire `devel/` est l'endroit où sont placés les fichiers générés lors de la compilation avant leur installation.


## ROS workspace défault

Vous pouvez avoir plusieurs Catkin workspace actifs simultanément.
ROS crée son propre Catkin workspace lors de l'installation. Dans ce tutoriel, nous utilisons **ROS Noetic**, dont l'espace de travail est installé au chemin `/opt/ros/noetic/share/`.

## Création d'un nouvel Catkin workspace

Pour créer un nouvel Catkin workspace, il suffit de créer l'espace source (`/src`). À l'aide de l'**EXPLORER** de fichiers situé dans le panneau de gauche, accédez au répertoire `packages/`. Vous constaterez que l'espace source, le sous-répertoire `/src`, a déjà été créé. Le Catkin workspace est donc prêt à l'emploi.


### Build workspace

Nous allons maintenant procéder à la compilation (ou build) de notre workspace nouvellement créé (et encore vide).

Pour ce faire, ouvrez un nouveau terminal (si cela n'avait pas déjà été fait) (`Menu -> Terminal -> Nouveau terminal`) puis exécutez les commandes suivantes :

```{bash}
    cd packages/
    catkin_make_isolated
```

**REMARQUE:** Il est important d'exécuter cette commande depuis le répertoire `packages/` (le répertoire de base du Catkin workspace).

**REMARQUE :** L'alternative habituelle à `catkin_make_isolated` est simplement `catkin make` ou `catkin build`. Dans ce cas, nous souhaitons forcer Catkin à compiler **uniquement** cet Catkin workspace. D'autres workspaces sont présents sur le système, comme nous le verrons plus tard.

Le processus de compilation devrait se terminer en quelques secondes et le résumé devrait indiquer qu'un seul package a été compilé. Il s'agit du package `catkin_tools_prebuild`, présent par défaut dans chaque nouvel espace de travail.


### Activer le workspace

Une fois le workspace est créé, nous souhaitons activer les artefacts générés. Pour ce faire, nous pouvons exécuter la commande suivante :

```{bash}
    source ./devel_isolated/setup.bash
```

Si tout se passe bien, cette commande devrait s'exécuter sans aucune erreur ni message d'erreur.
