# Catkin Packages

## Les composants d'un Catkin package?

Pour qu'un répertoire soit considéré comme un Catkin package, il doit répondre à plusieurs exigences :

- Il doit contenir un fichier `package.xml` conforme à Catkin, fournissant des métadonnées sur le package ;

- Il doit contenir un fichier `CMakeLists.txt` utilisant Catkin ;

- Il ne doit contenir qu'un seul package. Les packages imbriqués ou multiples partageant le même répertoire ne sont pas autorisés ;

Le package le plus simple possible pourrait avoir une structure similaire à celle-ci :

```
my_package/
  package.xml
  CMakeLists.txt
```


## Création d'un nouveau Catkin package

En utilisant l'**EXPLORER** situé sur le côté gauche de l'écran, vous pouvez constater que nous avons un répertoire nommé `dt-joystick-demo` dans l'espace source (`/src`) du workspace `./packages/src/` décrit précédemment. Il s'agit de notre package Catkin.


### Le ficher `package.xml`

Vous remarquerez un fichier nommé `package.xml` dans le répertoire `./packages/src/dt-joystick-demo/` dont le contenu est le suivant :

````python
<package>
  <name>dt-joystick-demo</name>
  <version>0.0.1</version>
  <description>
  My first Catkin package in Duckietown.
  </description>
  <maintainer email="YOUR_EMAIL@EXAMPLE.COM">YOUR_FULL_NAME</maintainer>
  <license>None</license>

  <buildtool_depend>catkin</buildtool_depend>
</package>
````

Vous pouvez remplacer `YOUR_FULL_NAME` par votre nom et prénom et `YOUR_EMAIL@EXAMPLE.COM` par votre adresse électronique.

### Le fichier `CMakeLists.txt`

Nous voyons également le fichier `CMakeLists.txt` dans le répertoire `./packages/src/dt-joystick-demo/` avec le contenu suivant :

````python
cmake_minimum_required(VERSION 2.8.3)
project(dt-joystick-demo)

find_package(catkin REQUIRED COMPONENTS
  rospy
)

catkin_package()
````

## Re-build Workspace

Nous pouvons maintenant procéder à la compilation (ou à la création) de notre workspace.

Pour ce faire, ouvrez un nouveau terminal (si nécessaire) (`Menu -> Terminal -> Nouveau terminal`) puis exécutez les commandes suivantes :

```{bash}
    cd packages/
    catkin_make_isolated
```

Le processus de compilation devrait se terminer en quelques secondes et le résumé devrait indiquer qu'un seul paquet a été compilé.


### Activer le workspace

Une fois le workspace est créé, nous souhaitons activer les artefacts générés. Pour ce faire, nous pouvons exécuter la commande suivante :

```{bash}
    source ./devel_isolated/setup.bash
```

Si tout se passe bien, cette commande devrait s'exécuter sans erreur ni message.

Nous pouvons vérifier que le nouveau package `dt-joystick-demo` a bien été reconnu par `catkin` en consultant la variable d'environnement `ROS_PACKAGE_PATH`, qui contient la liste des catkin packages actifs, séparés par deux-points.

Pour ce faire, exécutez la commande :

```{bash}
    echo $ROS_PACKAGE_PATH
```

Le résultat suivant devrait apparaître :

    /code/src/lx-ros-basics/packages/src/dt-joystick-demo:/code/src/dt-ros-commons/packages/duckietown:/code/src/dt-ros-commons/packages/duckietown_msgs:/code/src/dt-ros-commons/packages/duckietown_protocols:/opt/ros/noetic/share

Nous constatons que notre package `dt-joystick-demo` est bien le premier de la liste.
