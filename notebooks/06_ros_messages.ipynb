# Understanding ROS Messages

Ce tutoriel présente les messages ROS et explique comment créer des fichiers de définition de messages au format `.msg`.

## Introduction à `msg`

-   [msg](http://wiki.ros.org/msg): Les fichiers .msg sont de simples fichiers texte qui décrivent les champs d'un message ROS. Ils sont utilisés pour générer le code source des messages dans différents langages de programmation. Les fichiers .msg sont stockés dans le répertoire `msg` d'un package.

Les msg sont de simples fichiers texte comportant un type de champ et un nom de champ par ligne. Les types de champs que vous pouvez utiliser sont les suivants :

-   int8, int16, int32, int64 (plus uint\*)
-   float32, float64
-   string
-   time, duration
-   other msg files
-   variable-length array\[\] and fixed-length array\[C\]

Il existe également un type spécial dans ROS : `Header`. Cet `header` contient un `timestamp` et des informations sur le système de coordonnées, couramment utilisés dans ROS. Vous verrez fréquemment la première ligne d'un fichier .msg contenir un champ de type `Header` nommé `header`.

Voici un exemple de message qui utilise un `header`, une primitive de type `string` et deux autres messages :

```
  Header header
  string child_frame_id
  geometry_msgs/PoseWithCovariance pose
  geometry_msgs/TwistWithCovariance twist
```



## Utiliser msg

### Creation d'un msg

Définissons un nouveau message dans le package créé lors du tutoriel précédent. Dans l'**EXPLORER**, créez un nouveau package nommé `dt_joystick_demo_msgs`. Vous pouvez le faire en exécutant la commande suivante dans le répertoire `packages/src`:

```
catkin_create_pkg dt_joystick_demo_msgs rospy message_generation
```

Vous devriez constater que le dossier (package) `dt_joystick_demo_msgs` a été créé. À l'intérieur de ce dossier, créez un répertoire nommé `msg`.
Dans le dossier `msg`, créez un fichier nommé `Num.msg` avec le contenu suivant :

```
int64 num
```

Le fichier `.msg` présenté ci-dessus ne contient qu'une seule ligne. Vous pouvez bien sûr créer un fichier plus complexe en ajoutant plusieurs éléments, un par ligne, comme ceci :

```
string first_name
string last_name
uint8 age
uint32 score
```

Il reste cependant une étape. Nous devons nous assurer que les fichiers `.msg` soient convertis en code source pour le C++, Python et d'autres langages :

Ouvrez le fichier `package.xml` dans le package `dt_joystick_demo_msgs` et assurez-vous que cette ligne y figure et qu'elle n'est pas commentée (voir [cette page](http://www.htmlhelp.com/reference/wilbur/misc/comment.html) pour plus d'informations sur les commentaires).

```
  <exec_depend>message_runtime</exec_depend>
```

Notez qu'au moment de la compilation, nous avons besoin de `message_generation`, tandis qu'au moment de l'exécution, nous n'avons besoin que de `message_runtime`.

Ouvrez le fichier CMakeLists.txt dans le dossier `dt_joystick_demo_msgs`.

Ajoutez/décommentez le bloc de code suivant :

```
add_message_files(
  FILES
  Num.msg
)
```

En ajoutant les fichiers `.msg` manuellement, nous nous assurons que CMake sache quand il doit reconfigurer le projet après l'ajout d'autres fichiers `.msg`.

Nous devons maintenant nous assurer que la fonction `generate_messages()` est appelée en ajoutant ces lignes :

```
generate_messages()
```

Et enfin (quelque part après la ligne `generate_messages()`), modifiez la commande `catkin_package` pour y inclure `CATKIN_DEPENDS message_runtime` :


```
catkin_package(CATKIN_DEPENDS message_runtime)
```


Vous êtes maintenant prêt à générer les fichiers sources à partir de votre définition de message.

rosmsg show [message type]
```

Vous pouvez executer:

```
rosmsg show my_package/Num
```

Vous verrez :

-   ```
    int64 num
    ```
    

Dans l'exemple précédent, le type de message se compose de deux parties :

-   `my_package`: the package where the message is defined
    
-   `Num`: the name of the msg Num.
    

Si vous ne vous souvenez pas du package auquel appartient un message, vous pouvez omettre le nom du package. Essayez d'exécuter la commande suivante :

```
rosmsg show Num
```

Vous verrez :

-   ```
    [my_package/Num]:
    int64 num
    ```
    

Appliquons maintenant ce que nous avons appris sur ROS à Duckietown dans [le prochain notebook](../notebooks/07_duckietown_ros.ipynb).
